## Create purchase_level (Derived column)

```python
def purchase_level(value):
    if value < 5000:
        return 'Low'
    elif value < 10000:
        return 'Medium'
    else:
        return 'High'

df['Purchase_Level'] = df['Purchase'].apply(purchase_level)

df[['Purchase', 'Purchase_Level']].head()
```

A new Purchase_Level column is created using the rules: below 5,000 = Low; 5,000 to below 10,000 = Medium; 10,000 or more = High. The first five rows show the resulting labels alongside Purchase values.


## Overall Sales Analysis

```python
total_sales = df['Purchase'].sum()
average_purchase = df['Purchase'].mean()
minimum_purchase = df['Purchase'].min()
maximum_purchase = df['Purchase'].max()

print("Total Sales:", total_sales)
print("Average Purchase:", average_purchase)
print("Minimum Purchase:", minimum_purchase)
print("Maximum Purchase:", maximum_purchase)
```

Total sales are 5,095,812,742. The average transaction is about 9,263.97, with a minimum of 12 and a maximum of 23,961. These four measures provide the overall sales baseline.
